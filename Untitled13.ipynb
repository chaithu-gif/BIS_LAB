```python
import numpy as np

# 1. DEFINE YOUR MIXTURE PARAMETERS
N_COMPONENTS = 4     # Number of chemical ingredients
POP_SIZE = 50        # Number of candidate mixtures per generation
GENERATIONS = 100    # Total training iterations
MUTATION_RATE = 0.15 # Probability of mutating a mixture ratio

# 2. DEFINE THE FITNESS FUNCTION
def evaluate_mixture(individual):
    """
    Replace this mock formula with your true objective function.
    It can be an analytical equation, a machine learning model surrogate,
    or a lookup dictionary from lab experiments.
    """
    # Example: Components x0, x1, x2, x3 affect performance differently
    # Objective: Maximize performance value
    x0, x1, x2, x3 = individual
    performance_score = (x0 * 2.5) + (x1 * 6.0) - (x2 * 4.0) + (x3 * 1.5)
    return performance_score

# 3. INITIALIZE POPULATION
# Generate random ratios between 0 and 1
population = np.random.rand(POP_SIZE, N_COMPONENTS)
# Critical Step: Normalize rows so each candidate mixture sums to 1.0
population = population / population.sum(axis=1, keepdims=True)

print("Starting Optimization Optimization Loop...")

# 4. EVOLUTIONARY LOOP
for gen in range(GENERATIONS):
    # Calculate fitness scores for all candidate mixtures
    fitness_scores = np.array([evaluate_mixture(ind) for ind in population])

    # --- SELECTION (Roulette Wheel Selection) ---
    # Shift values to keep probabilities positive
    shifted_fitness = fitness_scores - np.min(fitness_scores) + 1e-6
    selection_probs = shifted_fitness / np.sum(shifted_fitness)
    selected_indices = np.random.choice(POP_SIZE, size=POP_SIZE, p=selection_probs)
    population = population[selected_indices]

    # --- CROSSOVER (Blend Recombination) ---
    for i in range(0, POP_SIZE, 2):
        if i + 1 < POP_SIZE:
            # Blend the proportions of two parents using a random weight alpha
            alpha = np.random.rand()
            child1 = alpha * population[i] + (1 - alpha) * population[i+1]
            child2 = (1 - alpha) * population[i] + alpha * population[i+1]
            population[i], population[i+1] = child1, child2

    # --- MUTATION ---
    for i in range(POP_SIZE):
        if np.random.rand() < MUTATION_RATE:
            # Introduce small variations into the proportions
            noise = np.random.normal(0, 0.05, N_COMPONENTS)
            population[i] = np.clip(population[i] + noise, 0, 1)

    # --- CONSTRAINT ENFORCEMENT ---
    # Re-normalize to fix any deviations introduced during crossover or mutation
    population = population / population.sum(axis=1, keepdims=True)

# 5. EXTRACT FINAL OPTIMAL MIXTURE
final_fitness = np.array([evaluate_mixture(ind) for ind in population])
best_index = np.argmax(final_fitness)
optimal_mixture = population[best_index]
best_score = final_fitness[best_index]

print("\n--- Optimization Complete ---")
print(f"Optimal Proportions: {np.round(optimal_mixture * 100, 2)} %")
print(f"Total Sum Check:      {np.sum(optimal_mixture):.1f}")
print(f"Maximized Property Value: {best_score:.4f}")
```
